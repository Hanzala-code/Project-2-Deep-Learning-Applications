**Requests per day for one worker (Little's law)**

L = lambda x W. With one worker, L = 1 and W = 0.056 s at my p50, so lambda = 1 / 0.056 = about 17.9 requests per second. Over a day that is 17.9 x 86,400 = about 1.5 million requests. This is an upper bound with the worker busy 100% of the time; real traffic arrives in peaks, so I would plan for a fraction of it.
